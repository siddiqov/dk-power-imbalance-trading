# First Extract Imbalance price Using API

```python
# Import necessary libraries
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# Machine Learning libraries
from sklearn.model_selection import train_test_split, GridSearchCV, TimeSeriesSplit
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import xgboost as xgb
import lightgbm as lgb

# Set random seed for reproducibility
np.random.seed(42)
```

```python
# ============================================
# PART 1: DATA EXTRACTION FROM API
# ============================================

def extract_imbalance_data(start_date=None, end_date=None, price_area=None, limit=10000):
    """
    Extract imbalance price data from Energidataservice API
    
    Parameters:
    - start_date: Start date for data extraction (YYYY-MM-DD)
    - end_date: End date for data extraction (YYYY-MM-DD)
    - price_area: DK1 or DK2
    - limit: Maximum number of records to fetch
    """
    
    base_url = "https://api.energidataservice.dk/dataset/ImbalancePrice"
    
    # Build query parameters
    params = {
        'limit': limit,
        'timezone': 'Europe/Copenhagen'
    }
    
    # Add date filters if provided
    if start_date:
        params['start'] = start_date
    if end_date:
        params['end'] = end_date
    if price_area:
        params['PriceArea'] = price_area
    
    try:
        # Make API request
        response = requests.get(base_url, params=params)
        response.raise_for_status()
        
        data = response.json()
        
        # Convert to DataFrame
        df = pd.DataFrame(data['records'])
        
        # Convert time columns to datetime
        if 'TimeUTC' in df.columns:
            df['TimeUTC'] = pd.to_datetime(df['TimeUTC'])
        if 'TimeDK' in df.columns:
            df['TimeDK'] = pd.to_datetime(df['TimeDK'])
        
        # Sort by time
        if 'TimeUTC' in df.columns:
            df = df.sort_values('TimeUTC')
        
        print(f"Successfully extracted {len(df)} records")
        if 'TimeUTC' in df.columns:
            print(f"Date range: {df['TimeUTC'].min()} to {df['TimeUTC'].max()}")
        if 'PriceArea' in df.columns:
            print(f"Price areas: {df['PriceArea'].unique()}")
        print(f"Columns: {df.columns.tolist()}")
        
        return df
        
    except requests.exceptions.RequestException as e:
        print(f"Error fetching data: {e}")
        return None

# Extract data for last 6 months (or as much as available)
end_date = datetime.now().strftime('%Y-%m-%d')
start_date = (datetime.now() - timedelta(days=180)).strftime('%Y-%m-%d')

print("Extracting data...")
df_raw = extract_imbalance_data(start_date=start_date, end_date=end_date, limit=10000)

if df_raw is not None:
    print(f"\nData shape: {df_raw.shape}")
    df_raw.head()
```

```python
# ============================================
# PART 2: DATA PREPROCESSING 
# ============================================

def preprocess_data(df):
    """
    Preprocess the imbalance price data with correct column names
    """
    df_copy = df.copy()
    
    # Show actual columns for debugging
    print("Available columns:", df_copy.columns.tolist())
    
    # Rename columns to more readable names
    rename_dict = {
        'TimeUTC': 'timestamp',
        'TimeDK': 'time_dk',
        'PriceArea': 'price_area',
        'SatisfiedDemand': 'satisfied_demand',
        'ImbalancePriceEUR': 'imbalance_price_eur',
        'ImbalancePriceDKK': 'imbalance_price_dkk',
        'SpotPriceEUR': 'spot_price_eur',
        'DominatingDirection': 'dominating_direction',
        'aFRRUpMW': 'afrr_up',
        'aFRRVWAUpEUR': 'afrr_vwa_up_eur',
        'aFRRVWAUpDKK': 'afrr_vwa_up_dkk',
        'aFRRDownMW': 'afrr_down',
        'aFRRVWADownEUR': 'afrr_vwa_down_eur',
        'aFRRVWADownDKK': 'afrr_vwa_down_dkk',
        'mFRRMarginalPriceUpEUR': 'mfrr_marginal_up_eur',
        'mFRRMarginalPriceUpDKK': 'mfrr_marginal_up_dkk',
        'mFRRMarginalPriceDownEUR': 'mfrr_marginal_down_eur',
        'mFRRMarginalPriceDownDKK': 'mfrr_marginal_down_dkk'
    }
    
    # Rename columns that exist
    rename_dict_filtered = {k: v for k, v in rename_dict.items() if k in df_copy.columns}
    df_processed = df_copy.rename(columns=rename_dict_filtered)
    
    # Ensure timestamp is datetime
    if 'timestamp' in df_processed.columns:
        df_processed['timestamp'] = pd.to_datetime(df_processed['timestamp'])
        
        # Create time-based features
        df_processed['hour'] = df_processed['timestamp'].dt.hour
        df_processed['day_of_week'] = df_processed['timestamp'].dt.dayofweek
        df_processed['month'] = df_processed['timestamp'].dt.month
        df_processed['day_of_year'] = df_processed['timestamp'].dt.dayofyear
    
    # Handle missing values for numeric columns
    numeric_cols = ['satisfied_demand', 'imbalance_price_eur', 'spot_price_eur', 
                    'afrr_up', 'afrr_down']
    for col in numeric_cols:
        if col in df_processed.columns:
            df_processed[col] = pd.to_numeric(df_processed[col], errors='coerce')
            df_processed[col] = df_processed[col].fillna(df_processed[col].median())
    
    # Handle dominating direction (categorical)
    if 'dominating_direction' in df_processed.columns:
        df_processed['dominating_direction'] = df_processed['dominating_direction'].fillna(0)
        # Convert to numeric if it's categorical
        df_processed['dominating_direction'] = pd.to_numeric(df_processed['dominating_direction'], errors='coerce').fillna(0)
    else:
        df_processed['dominating_direction'] = 0
    
    # Use Imbalance Price (EUR) as target
    if 'imbalance_price_eur' in df_processed.columns:
        df_processed['imbalance_price'] = df_processed['imbalance_price_eur']
    else:
        # If EUR price not available, use DKK price or create a placeholder
        if 'imbalance_price_dkk' in df_processed.columns:
            df_processed['imbalance_price'] = df_processed['imbalance_price_dkk'] / 7.46  # Approximate DKK to EUR
        else:
            df_processed['imbalance_price'] = 0
    
    # Use Spot Price (EUR) as reference
    if 'spot_price_eur' in df_processed.columns:
        df_processed['spot_price'] = df_processed['spot_price_eur']
    else:
        df_processed['spot_price'] = 0
    
    # Create lag features for imbalance price
    for lag in [1, 2, 3, 6, 12, 24]:
        df_processed[f'lag_{lag}_imbalance'] = df_processed.groupby('price_area')['imbalance_price'].shift(lag)
    
    # Create rolling statistics
    for window in [3, 6, 12, 24]:
        df_processed[f'rolling_mean_{window}'] = df_processed.groupby('price_area')['imbalance_price'].transform(
            lambda x: x.rolling(window=window, min_periods=1).mean()
        )
        df_processed[f'rolling_std_{window}'] = df_processed.groupby('price_area')['imbalance_price'].transform(
            lambda x: x.rolling(window=window, min_periods=1).std()
        )
    
    # Create price difference features
    df_processed['price_diff_spot_imbalance'] = df_processed['spot_price'] - df_processed['imbalance_price']
    df_processed['price_ratio_spot_imbalance'] = df_processed['spot_price'] / (df_processed['imbalance_price'] + 1e-5)
    
    # Create regime indicator (positive/negative imbalance)
    df_processed['regime'] = np.where(df_processed['imbalance_price'] > df_processed['spot_price'], 1, 0)
    
    return df_processed

# Preprocess the data
df_processed = preprocess_data(df_raw)

print("\nProcessed data shape:", df_processed.shape)
print("\nProcessed data columns:", df_processed.columns.tolist())
df_processed.head()
```

```python
# ============================================
# PART 4: PREPARE DATA FOR MODELING
# ============================================

def prepare_model_data(df, target_col='imbalance_price', test_size=0.2):
    """
    Prepare data for modeling with time series split
    """
    
    # Remove rows with NaN in target or features
    df_clean = df.dropna(subset=[target_col])
    
    # Define features - only use columns that exist
    feature_cols = ['satisfied_demand', 'spot_price', 'hour', 'day_of_week', 'month']
    
    # Add other features if they exist
    if 'dominating_direction' in df_clean.columns:
        feature_cols.append('dominating_direction')
    if 'afrr_up' in df_clean.columns:
        feature_cols.append('afrr_up')
    if 'afrr_down' in df_clean.columns:
        feature_cols.append('afrr_down')
    if 'price_diff_spot_imbalance' in df_clean.columns:
        feature_cols.append('price_diff_spot_imbalance')
    if 'price_ratio_spot_imbalance' in df_clean.columns:
        feature_cols.append('price_ratio_spot_imbalance')
    
    # Add additional price features if available
    if 'imbalance_price_dkk' in df_clean.columns:
        feature_cols.append('imbalance_price_dkk')
    if 'afrr_vwa_up_eur' in df_clean.columns:
        feature_cols.append('afrr_vwa_up_eur')
    if 'afrr_vwa_down_eur' in df_clean.columns:
        feature_cols.append('afrr_vwa_down_eur')
    
    # Add lag features
    lag_cols = [col for col in df_clean.columns if 'lag_' in col]
    feature_cols.extend(lag_cols)
    
    # Add rolling features
    roll_cols = [col for col in df_clean.columns if 'rolling_' in col]
    feature_cols.extend(roll_cols)
    
    # Remove any duplicates and ensure all columns exist
    feature_cols = list(set(feature_cols))
    feature_cols = [col for col in feature_cols if col in df_clean.columns]
    
    print(f"Using {len(feature_cols)} features: {feature_cols}")
    
    # Prepare X and y
    X = df_clean[feature_cols].copy()
    y = df_clean[target_col].copy()
    
    # Handle categorical variables
    if 'dominating_direction' in X.columns:
        X['dominating_direction'] = X['dominating_direction'].astype('category')
        X = pd.get_dummies(X, columns=['dominating_direction'], drop_first=True)
    
    # Split by time (not random)
    train_size = int(len(X) * (1 - test_size))
    
    X_train = X[:train_size]
    X_test = X[train_size:]
    y_train = y[:train_size]
    y_test = y[train_size:]
    
    # Scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    feature_names = X_train.columns.tolist()
    
    print(f"Training set size: {len(X_train)}")
    print(f"Test set size: {len(X_test)}")
    print(f"Number of features: {len(feature_names)}")
    
    return X_train_scaled, X_test_scaled, y_train, y_test, scaler, feature_names

# Prepare data for DK1 and DK2 separately
models_results = {}

# Check if we have price_area column
if 'price_area' in df_processed.columns:
    areas = df_processed['price_area'].unique()
else:
    areas = ['All']

for area in areas:
    print("\n" + "="*50)
    print(f"Preparing data for {area}")
    print("="*50)
    if area != 'All':
        area_data = df_processed[df_processed['price_area'] == area].copy()
    else:
        area_data = df_processed.copy()
    
    # Only proceed if we have enough data
    if len(area_data) > 100:
        X_train, X_test, y_train, y_test, scaler, features = prepare_model_data(
            area_data, test_size=0.2
        )
        
        models_results[area] = {
            'X_train': X_train,
            'X_test': X_test,
            'y_train': y_train,
            'y_test': y_test,
            'scaler': scaler,
            'features': features,
            'area_data': area_data
        }
    else:
        print(f"Not enough data for {area} (only {len(area_data)} records)")
```

```python
# ============================================
# PART 5: MODEL DEFINITION WITH GRID SEARCH
# ============================================

def create_models_simple():
    """
    Create models with simplified hyperparameter grids for faster execution
    """
    
    models = {
        'Random Forest': {
            'model': RandomForestRegressor(random_state=42, n_jobs=-1),
            'param_grid': {
                'n_estimators': [50, 100],
                'max_depth': [10, 20],
                'min_samples_split': [2, 5]
            }
        },
        'XGBoost': {
            'model': xgb.XGBRegressor(random_state=42, n_jobs=-1),
            'param_grid': {
                'n_estimators': [50, 100],
                'max_depth': [3, 5],
                'learning_rate': [0.05, 0.1]
            }
        },
        'LightGBM': {
            'model': lgb.LGBMRegressor(random_state=42, n_jobs=-1, verbose=-1),
            'param_grid': {
                'n_estimators': [50, 100],
                'max_depth': [3, 5],
                'learning_rate': [0.05, 0.1]
            }
        }
    }
    
    return models
```

```python
# ============================================
# PART 6: TRAIN MODELS WITH GRID SEARCH
# ============================================

def train_models_with_gridsearch(X_train, y_train, models, cv=3):
    """
    Train models using grid search with time series cross-validation
    """
    
    results = {}
    
    for model_name, model_info in models.items():
        print(f"\nTraining {model_name}...")
        
        try:
            # Time series cross-validation
            tscv = TimeSeriesSplit(n_splits=cv)
            
            grid_search = GridSearchCV(
                model_info['model'],
                model_info['param_grid'],
                cv=tscv,
                scoring='neg_mean_squared_error',
                n_jobs=-1,
                verbose=0
            )
            
            grid_search.fit(X_train, y_train)
            
            results[model_name] = {
                'best_model': grid_search.best_estimator_,
                'best_params': grid_search.best_params_,
                'best_score': -grid_search.best_score_,
                'cv_results': grid_search.cv_results_
            }
            
            print(f"Best parameters: {grid_search.best_params_}")
            print(f"Best CV RMSE: {np.sqrt(-grid_search.best_score_):.4f}")
            
        except Exception as e:
            print(f"Error training {model_name}: {e}")
            results[model_name] = None
    
    return results
```

```python
# ============================================
# PART 7: EVALUATE AND COMPARE MODELS
# ============================================

def evaluate_models(models_results, X_test, y_test, scaler=None, features=None):
    """
    Evaluate trained models on test data
    """
    
    evaluation = {}
    
    for model_name, model_info in models_results.items():
        if model_info is None:
            continue
            
        best_model = model_info['best_model']
        
        # Predict on test data
        y_pred = best_model.predict(X_test)
        
        # Calculate metrics
        mae = mean_absolute_error(y_test, y_pred)
        rmse = np.sqrt(mean_squared_error(y_test, y_pred))
        r2 = r2_score(y_test, y_pred)
        
        # Calculate MAPE
        mape = np.mean(np.abs((y_test - y_pred) / (y_test + 1e-5))) * 100
        
        evaluation[model_name] = {
            'MAE': mae,
            'RMSE': rmse,
            'R2': r2,
            'MAPE': mape,
            'predictions': y_pred,
            'best_params': model_info['best_params']
        }
        
        print(f"\n{model_name}:")
        print(f"  MAE: {mae:.4f}")
        print(f"  RMSE: {rmse:.4f}")
        print(f"  R²: {r2:.4f}")
        print(f"  MAPE: {mape:.2f}%")
    
    return evaluation

# Function to plot predictions
def plot_predictions(y_test, predictions_dict, title="Model Predictions"):
    """
    Plot actual vs predicted values
    """
    
    n_models = len(predictions_dict)
    if n_models == 0:
        print("No predictions to plot")
        return
    
    fig, axes = plt.subplots(n_models, 1, figsize=(12, 4*n_models))
    
    if n_models == 1:
        axes = [axes]
    
    for idx, (model_name, y_pred) in enumerate(predictions_dict.items()):
        # Plot only first 200 points for clarity
        sample_size = min(200, len(y_test))
        
        axes[idx].plot(y_test[:sample_size].values, label='Actual', alpha=0.7)
        axes[idx].plot(y_pred[:sample_size], label='Predicted', alpha=0.7)
        axes[idx].set_title(f"{model_name} - Predictions vs Actual")
        axes[idx].set_xlabel('Time Steps')
        axes[idx].set_ylabel('Imbalance Price (EUR)')
        axes[idx].legend()
        axes[idx].grid(True)
    
    plt.tight_layout()
    plt.show()
```

```python
# ============================================
# PART 8: RUN MODELS FOR BOTH AREAS
# ============================================

# Choose model set (simplified for faster execution)
models = create_models_simple()

# Store all results
all_results = {}

for area in models_results.keys():
    print(f"\n{'='*60}")
    print(f"MODEL TRAINING FOR {area}")
    print('='*60)
    
    data = models_results[area]
    
    # Train models with grid search
    trained_models = train_models_with_gridsearch(
        data['X_train'], 
        data['y_train'], 
        models,
        cv=3
    )
    
    # Evaluate models
    print(f"\n{'='*50}")
    print(f"EVALUATION FOR {area}")
    print('='*50)
    
    evaluation = evaluate_models(
        trained_models,
        data['X_test'],
        data['y_test']
    )
    
    # Store results
    all_results[area] = {
        'trained_models': trained_models,
        'evaluation': evaluation,
        'data': data
    }
    
    # Plot predictions
    predictions_dict = {}
    for model_name, eval_info in evaluation.items():
        predictions_dict[model_name] = eval_info['predictions']
    
    if predictions_dict:
        plot_predictions(
            data['y_test'], 
            predictions_dict, 
            title=f"{area} - Model Predictions Comparison"
        )
```

```python
# ============================================
# PART 9: MODEL COMPARISON AND ANALYSIS
# ============================================

def compare_models(all_results):
    """
    Compare model performance across areas and methods
    """
    
    comparison_df = []
    
    for area, results in all_results.items():
        if 'evaluation' in results:
            for model_name, eval_info in results['evaluation'].items():
                comparison_df.append({
                    'Price Area': area,
                    'Model': model_name,
                    'MAE': eval_info['MAE'],
                    'RMSE': eval_info['RMSE'],
                    'R2': eval_info['R2'],
                    'MAPE': eval_info['MAPE']
                })
    
    if not comparison_df:
        print("No comparison data available")
        return pd.DataFrame()
    
    comparison_df = pd.DataFrame(comparison_df)
    
    # Display comparison table
    print("\n" + "="*70)
    print("MODEL PERFORMANCE COMPARISON")
    print("="*70)
    
    for area in comparison_df['Price Area'].unique():
        print(f"\n{area}:")
        print(comparison_df[comparison_df['Price Area'] == area].to_string(index=False))
    
    # Visualization
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    # MAE comparison
    sns.barplot(data=comparison_df, x='Model', y='MAE', hue='Price Area', ax=axes[0,0])
    axes[0,0].set_title('Mean Absolute Error (MAE)')
    axes[0,0].grid(True)
    
    # RMSE comparison
    sns.barplot(data=comparison_df, x='Model', y='RMSE', hue='Price Area', ax=axes[0,1])
    axes[0,1].set_title('Root Mean Squared Error (RMSE)')
    axes[0,1].grid(True)
    
    # R2 comparison
    sns.barplot(data=comparison_df, x='Model', y='R2', hue='Price Area', ax=axes[1,0])
    axes[1,0].set_title('R-squared Score')
    axes[1,0].grid(True)
    
    # MAPE comparison
    sns.barplot(data=comparison_df, x='Model', y='MAPE', hue='Price Area', ax=axes[1,1])
    axes[1,1].set_title('Mean Absolute Percentage Error (MAPE)')
    axes[1,1].grid(True)
    
    plt.tight_layout()
    plt.show()
    
    return comparison_df

# Only run comparison if we have results
if all_results:
    comparison_df = compare_models(all_results)
else:
    print("No models were trained successfully. Skipping comparison.")
    comparison_df = pd.DataFrame()
```

```python
# ============================================
# PART 10: SAVE RESULTS TO CSV
# ============================================

def save_results(df_processed, all_results, comparison_df):
    """
    Save processed data and results to CSV files
    """
    
    # Save processed data
    df_processed.to_csv('imbalance_price_processed.csv', index=False)
    print("✓ Saved processed data to 'imbalance_price_processed.csv'")
    
    # Save comparison results
    if not comparison_df.empty:
        comparison_df.to_csv('model_comparison_results.csv', index=False)
        print("✓ Saved model comparison to 'model_comparison_results.csv'")
    
    # Save detailed predictions
    for area, results in all_results.items():
        if 'data' in results and 'y_test' in results['data'] and 'area_data' in results['data']:
            predictions_df = pd.DataFrame({
                'Actual': results['data']['y_test'].values,
                'Timestamp': results['data']['area_data'].iloc[
                    len(results['data']['y_train']):
                ]['timestamp'].values
            })
            
            for model_name, eval_info in results['evaluation'].items():
                predictions_df[f'Predicted_{model_name}'] = eval_info['predictions']
            
            predictions_df.to_csv(f'predictions_{area}.csv', index=False)
            print(f"✓ Saved predictions for {area} to 'predictions_{area}.csv'")
    
    # Save model parameters
    params_list = []
    for area, results in all_results.items():
        for model_name, model_info in results['trained_models'].items():
            if model_info is not None:
                params_list.append({
                    'Price Area': area,
                    'Model': model_name,
                    'Best Parameters': str(model_info['best_params']),
                    'CV Score': model_info['best_score']
                })
    
    if params_list:
        params_df = pd.DataFrame(params_list)
        params_df.to_csv('model_parameters.csv', index=False)
        print("✓ Saved model parameters to 'model_parameters.csv'")

# Save all results
print("\n" + "="*50)
print("SAVING RESULTS")
print("="*50)
save_results(df_processed, all_results, comparison_df)
```

```python
# ============================================
# PART 11: REGIME SWITCH PREDICTION (CLASSIFICATION)
# ============================================

def prepare_regime_data(df, target_col='regime', test_size=0.2):
    """
    Prepare data for regime switch prediction (classification)
    """
    
    # Define features
    feature_cols = ['satisfied_demand', 'spot_price', 'hour', 'day_of_week', 'month']
    
    if 'dominating_direction' in df.columns:
        feature_cols.append('dominating_direction')
    if 'afrr_up' in df.columns:
        feature_cols.append('afrr_up')
    if 'afrr_down' in df.columns:
        feature_cols.append('afrr_down')
    if 'price_diff_spot_imbalance' in df.columns:
        feature_cols.append('price_diff_spot_imbalance')
    if 'price_ratio_spot_imbalance' in df.columns:
        feature_cols.append('price_ratio_spot_imbalance')
    
    # Add lag features
    lag_cols = [col for col in df.columns if 'lag_' in col]
    feature_cols.extend(lag_cols)
    
    # Add rolling features
    roll_cols = [col for col in df.columns if 'rolling_' in col]
    feature_cols.extend(roll_cols)
    
    # Remove duplicates and ensure columns exist
    feature_cols = list(set(feature_cols))
    feature_cols = [col for col in feature_cols if col in df.columns]
    
    # Prepare features
    X = df[feature_cols].copy()
    
    # Handle categorical variables
    if 'dominating_direction' in X.columns:
        X['dominating_direction'] = X['dominating_direction'].astype('category')
        X = pd.get_dummies(X, columns=['dominating_direction'], drop_first=True)
    
    y = df[target_col].copy()
    
    # Split data
    train_size = int(len(X) * (1 - test_size))
    
    X_train = X[:train_size]
    X_test = X[train_size:]
    y_train = y[:train_size]
    y_test = y[train_size:]
    
    # Scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    return X_train_scaled, X_test_scaled, y_train, y_test, scaler

print("\n" + "="*50)
print("REGIME SWITCH PREDICTION (Classification)")
print("="*50)

regime_results = {}

for area in models_results.keys():
    print(f"\n{'='*50}")
    print(f"Training Regime Classifier for {area}")
    print('='*50)
    
    area_data = models_results[area]['area_data']
    
    # Prepare data
    X_train, X_test, y_train, y_test, scaler = prepare_regime_data(area_data)
    
    # Define models for classification
    classifiers = {
        'Random Forest': RandomForestClassifier(random_state=42, n_jobs=-1),
        'Gradient Boosting': GradientBoostingClassifier(random_state=42),
        'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000)
    }
    
    # Train and evaluate each classifier
    regime_results[area] = {}
    
    for clf_name, clf in classifiers.items():
        print(f"\nTraining {clf_name}...")
        
        # Grid search for classifiers
        if clf_name == 'Random Forest':
            param_grid = {
                'n_estimators': [50, 100],
                'max_depth': [5, 10],
                'min_samples_split': [2, 5]
            }
        elif clf_name == 'Gradient Boosting':
            param_grid = {
                'n_estimators': [50, 100],
                'learning_rate': [0.05, 0.1],
                'max_depth': [3, 5]
            }
        else:  # Logistic Regression
            param_grid = {
                'C': [0.1, 1, 10]
            }
        
        try:
            grid_search = GridSearchCV(
                clf, param_grid, cv=3, scoring='accuracy', n_jobs=-1
            )
            grid_search.fit(X_train, y_train)
            
            # Best model
            best_clf = grid_search.best_estimator_
            
            # Predictions
            y_pred = best_clf.predict(X_test)
            
            # Metrics
            accuracy = accuracy_score(y_test, y_pred)
            conf_matrix = confusion_matrix(y_test, y_pred)
            
            print(f"Best parameters: {grid_search.best_params_}")
            print(f"Accuracy: {accuracy:.4f}")
            print("\nClassification Report:")
            print(classification_report(y_test, y_pred, target_names=['Normal', 'Regime Switch']))
            
            regime_results[area][clf_name] = {
                'model': best_clf,
                'accuracy': accuracy,
                'confusion_matrix': conf_matrix,
                'predictions': y_pred
            }
        except Exception as e:
            print(f"Error training {clf_name}: {e}")
```

```python
# ============================================
# PART 12: REGIME SWITCH VISUALIZATION (COMPLETE FIX)
# ============================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_regime_results(regime_results, df_processed):
    """
    Visualize regime switch predictions
    """
    
    if not regime_results:
        print("No regime results to plot")
        return
    
    for area in regime_results.keys():
        if area not in regime_results or not regime_results[area]:
            continue
        
        # Create a 2x2 subplot grid
        fig = plt.figure(figsize=(15, 10))
        
        # Get data for this area
        if 'price_area' in df_processed.columns:
            area_data = df_processed[df_processed['price_area'] == area]
        else:
            area_data = df_processed
        
        # Plot 1: Price with regime indicators (top left)
        ax1 = plt.subplot(2, 2, 1)
        if 'timestamp' in area_data.columns and 'imbalance_price' in area_data.columns:
            ax1.plot(area_data['timestamp'], area_data['imbalance_price'], 
                    label='Imbalance Price', alpha=0.7)
            
            # Add regime indicators
            if 'regime' in area_data.columns:
                regime_points = area_data[area_data['regime'] == 1]
                ax1.scatter(regime_points['timestamp'], regime_points['imbalance_price'],
                          color='red', label='Regime Switch', s=10, alpha=0.5)
            
            ax1.set_title(f'{area} - Imbalance Price with Regime Switches')
            ax1.set_xlabel('Date')
            ax1.set_ylabel('Price (EUR)')
            ax1.legend()
            ax1.grid(True)
        else:
            ax1.text(0.5, 0.5, 'No price data available', 
                    horizontalalignment='center', verticalalignment='center')
            ax1.set_title(f'{area} - No Price Data')
        
        # Plot 2: Regime distribution (top right)
        ax2 = plt.subplot(2, 2, 2)
        if 'regime' in area_data.columns:
            regime_dist = area_data['regime'].value_counts()
            ax2.pie(regime_dist.values, labels=['Normal', 'Regime Switch'], 
                   autopct='%1.1f%%')
        else:
            ax2.text(0.5, 0.5, 'No regime data available', 
                    horizontalalignment='center', verticalalignment='center')
        ax2.set_title(f'{area} - Regime Distribution')
        
        # Plot 3: Model comparison (bottom left)
        ax3 = plt.subplot(2, 2, 3)
        if regime_results[area]:
            models = list(regime_results[area].keys())
            accuracies = [regime_results[area][model]['accuracy'] for model in models]
            
            ax3.bar(models, accuracies, color='skyblue')
            ax3.set_title(f'{area} - Model Comparison')
            ax3.set_xlabel('Model')
            ax3.set_ylabel('Accuracy')
            ax3.set_ylim([0, 1])
            ax3.grid(True)
            
            # Add accuracy values on bars
            for i, v in enumerate(accuracies):
                ax3.text(i, v + 0.02, f'{v:.3f}', ha='center', va='bottom')
        else:
            ax3.text(0.5, 0.5, 'No model results available', 
                    horizontalalignment='center', verticalalignment='center')
            ax3.set_title(f'{area} - No Model Results')
        
        # Plot 4: Confusion matrix (bottom right)
        ax4 = plt.subplot(2, 2, 4)
        if regime_results[area]:
            # Show the best model's confusion matrix
            best_model = max(regime_results[area].items(), key=lambda x: x[1]['accuracy'])
            model_name = best_model[0]
            cm = best_model[1]['confusion_matrix']
            
            sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax4)
            ax4.set_title(f'{model_name} - Confusion Matrix')
            ax4.set_xlabel('Predicted')
            ax4.set_ylabel('Actual')
            ax4.set_xticklabels(['Normal', 'Switch'])
            ax4.set_yticklabels(['Normal', 'Switch'])
        else:
            ax4.text(0.5, 0.5, 'No confusion matrix available', 
                    horizontalalignment='center', verticalalignment='center')
            ax4.set_title(f'{area} - No Confusion Matrix')
        
        plt.tight_layout()
        plt.show()

# Plot regime results if available
if 'regime_results' in locals() and regime_results:
    plot_regime_results(regime_results, df_processed)
else:
    print("\n⚠ No regime results available to plot")
```

```python
# ============================================
# PART 13: FEATURE IMPORTANCE ANALYSIS
# ============================================

def analyze_feature_importance(trained_models, features, area):
    """
    Analyze feature importance for tree-based models
    """
    
    print(f"\n{'='*50}")
    print(f"FEATURE IMPORTANCE FOR {area}")
    print('='*50)
    
    feature_importance_df = pd.DataFrame()
    
    for model_name, model_info in trained_models.items():
        if model_info is None:
            continue
        
        model = model_info['best_model']
        
        # Check if model has feature_importances_ attribute
        if hasattr(model, 'feature_importances_'):
            importances = model.feature_importances_
            
            # Ensure we have the right number of features
            n_features = min(len(importances), len(features))
            
            # Create DataFrame
            temp_df = pd.DataFrame({
                'Feature': features[:n_features],
                'Importance': importances[:n_features],
                'Model': model_name
            })
            
            temp_df = temp_df.sort_values('Importance', ascending=False)
            
            print(f"\n{model_name} - Top 10 Features:")
            print(temp_df.head(10).to_string(index=False))
            
            feature_importance_df = pd.concat([feature_importance_df, temp_df])
    
    return feature_importance_df

# Analyze feature importance for each area
if all_results:
    all_feature_importance = {}
    
    for area in all_results.keys():
        if 'trained_models' in all_results[area] and 'data' in all_results[area]:
            importance_df = analyze_feature_importance(
                all_results[area]['trained_models'],
                all_results[area]['data']['features'],
                area
            )
            all_feature_importance[area] = importance_df

    # Plot feature importance
    def plot_feature_importance(importance_df, area):
        """
        Plot feature importance for tree-based models
        """
        
        if importance_df.empty:
            print(f"No feature importance data for {area}")
            return
        
        fig, axes = plt.subplots(1, 2, figsize=(15, 5))
        
        models = importance_df['Model'].unique()
        for idx, model_name in enumerate(models[:2]):
            model_data = importance_df[importance_df['Model'] == model_name].head(10)
            
            ax = axes[idx]
            ax.barh(model_data['Feature'], model_data['Importance'], color='skyblue')
            ax.set_title(f'{area} - {model_name} Top 10 Features')
            ax.set_xlabel('Importance')
            ax.set_ylabel('Feature')
            ax.grid(True)
        
        plt.tight_layout()
        plt.show()

    for area in all_feature_importance.keys():
        plot_feature_importance(all_feature_importance[area], area)
else:
    print("\n⚠ No models available for feature importance analysis")
```
